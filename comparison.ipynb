```python
import matplotlib.pyplot as plt
import pandas as pd

# --- Fixed results table ---
data = [
    ["MediaPipe",          24.56, 12.772, 0.61, 57.66],
    ["YOLOv11-default",    20.17, 44.055, 0.34, 61.38],
    ["NeuroYOLO-finetuned",17.57,  4.311, 0.13, 57.86],
    ["YOLOv8n-pose",       78.82,  7.897, 0.33, 58.54],
    ["MoveNet-Thunder",    24.08, 15.370, 1.10, 17.50],
]

df = pd.DataFrame(data, columns=["Model", "ms/frame", "Jitter Var", "MSI", "PSM"])

# --- Sort models by runtime (optional) ---
df = df.sort_values("ms/frame")
```

```python
# --- Plot 1: Inference efficiency ---
plt.figure(figsize=(7,4))
plt.barh(df["Model"], df["ms/frame"], color="lightsteelblue", edgecolor="black")
plt.xlabel("Inference Time (ms/frame)")
plt.title("Model Efficiency Comparison")
plt.grid(axis='x', linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
# --- Plot 2: Temporal stability (Jitter Variance) ---
plt.figure(figsize=(7,4))
plt.barh(df["Model"], df["Jitter Var"], color="lightcoral", edgecolor="black")
plt.xlabel("Temporal Jitter Variance")
plt.title("Temporal Stability Comparison")
plt.grid(axis='x', linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
# --- Plot 3: Movement smoothness (MSI) and Pose Similarity (PSM) ---
fig, ax1 = plt.subplots(figsize=(7,4))
ax2 = ax1.twiny()
ax1.barh(df["Model"], df["MSI"], color="palegreen", edgecolor="black", label="MSI ↓")
ax2.plot(df["PSM"], df["Model"], "o-", color="royalblue", label="PSM ↑")

ax1.set_xlabel("Movement Smoothness Index (lower is smoother)")
ax2.set_xlabel("Pose Similarity Metric (higher is better)")
plt.title("Biomechanical Quality Metrics Across Models")
ax1.grid(axis='x', linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Data
data = [
    ["MediaPipe",          24.56, 12.772, 0.61, 57.66],
    ["YOLOv11-default",    20.17, 44.055, 0.34, 61.38],
    ["NeuroYOLO-finetuned",17.57,  4.311, 0.13, 57.86],
    ["YOLOv8n-pose",       78.82,  7.897, 0.33, 58.54],
    ["MoveNet-Thunder",    24.08, 15.370, 1.10, 17.50],
]

df = pd.DataFrame(data, columns=["Model", "ms/frame", "Jitter Var", "MSI", "PSM"])
df = df.sort_values("ms/frame")

# Shared style
plt.rcParams.update({
    "font.size": 11,
    "font.family": "DejaVu Sans",
    "figure.dpi": 300,
    "axes.edgecolor": "0.3",
    "axes.linewidth": 0.8,
    "grid.alpha": 0.4
})

# Figure 1 – Efficiency
plt.figure(figsize=(5,3))
plt.barh(df["Model"], df["ms/frame"], color="#9EC3E6", edgecolor="0.3")
plt.xlabel("Inference Time (ms/frame)")
plt.title("Model Efficiency Comparison", pad=10)
plt.grid(axis='x', linestyle="--")
plt.tight_layout()
plt.savefig("Fig_efficiency.png", bbox_inches="tight")
plt.show()

# Figure 2 – Temporal Stability
plt.figure(figsize=(5,3))
plt.barh(df["Model"], df["Jitter Var"], color="#F29C9C", edgecolor="0.3")
plt.xlabel("Temporal Jitter Variance")
plt.title("Temporal Stability Comparison", pad=10)
plt.grid(axis='x', linestyle="--")
plt.tight_layout()
plt.savefig("Fig_stability.png", bbox_inches="tight")
plt.show()

# Figure 3 – Biomechanical Metrics
fig, ax1 = plt.subplots(figsize=(5,3))
ax2 = ax1.twiny()

ax1.barh(df["Model"], df["MSI"], color="#A9ECA2", edgecolor="0.3", label="MSI ↓")
ax2.plot(df["PSM"], df["Model"], "o-", color="#386CB0", label="PSM ↑")

ax1.set_xlabel("Movement Smoothness Index (lower is smoother)")
ax2.set_xlabel("Pose Similarity Metric (higher is better)")
ax1.set_title("Biomechanical Quality Metrics Across Models", pad=10)
ax1.grid(axis='x', linestyle="--")

# Align legends
lines, labels = [], []
for ax in [ax1, ax2]:
    L, l = ax.get_legend_handles_labels()
    lines += L; labels += l
ax1.legend(lines, labels, loc="lower right")

plt.tight_layout()
plt.savefig("Fig_biomech_metrics.png", bbox_inches="tight")
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# --- Data ---
data = {
    "Model": [
        "MediaPipe",
        "MoveNet-Thunder",
        "NeuroYOLO-finetuned",
        "YOLOv11-default",
        "YOLOv8n-pose"
    ],
    "Jitter Var": [12.772, 15.370, 4.311, 44.055, 7.897],
    "MSI": [0.61, 1.10, 0.13, 0.34, 0.33],
    "PSM": [57.66, 17.50, 57.86, 61.38, 58.54]
}

df = pd.DataFrame(data)

# --- Aesthetic setup ---
plt.rcParams.update({
    "figure.figsize": (10, 3.5),
    "font.size": 11,
    "axes.edgecolor": "0.3",
    "axes.linewidth": 0.8,
    "grid.alpha": 0.4,
    "axes.titlesize": 11
})

colors = ["#9EC3E6", "#F4A582", "#A9ECA2", "#F46D43", "#386CB0"]

# --- Create subplots ---
fig, axes = plt.subplots(1, 3, figsize=(10, 3.5))
fig.suptitle("Comparative Evaluation of Pose Estimation Models", fontsize=13, y=1.03)

# --- Subplot 1: Jitter variance ---
axes[0].bar(df["Model"], df["Jitter Var"], color=colors, edgecolor="black")
axes[0].set_title("Keypoint Jitter Variance (px²) ↓")
axes[0].grid(axis="y", linestyle="--", alpha=0.5)

# --- Subplot 2: Smoothness index ---
axes[1].bar(df["Model"], df["MSI"], color=colors, edgecolor="black")
axes[1].set_title("Movement Smoothness Index ↓")
axes[1].grid(axis="y", linestyle="--", alpha=0.5)

# --- Subplot 3: Pose similarity ---
axes[2].bar(df["Model"], df["PSM"], color=colors, edgecolor="black")
axes[2].set_title("Pose Similarity (%) ↑")
axes[2].grid(axis="y", linestyle="--", alpha=0.5)

# --- Formatting ---
for ax in axes:
    ax.set_xticklabels(df["Model"], rotation=30, ha="right")

plt.tight_layout()
plt.savefig("Fig_comparative_models.png", dpi=300, bbox_inches="tight")
plt.show()
```
